```python
import json
import os
from google.colab import drive
drive.mount('/content/drive')

TASKS_FILE = "/content/drive/My Drive/tasks.json"

def load_tasks():
    """Load tasks from the JSON file. Return empty list if file doesn't exist."""
    if os.path.exists(TASKS_FILE):
        try:
            with open(TASKS_FILE, "r") as file:
                return json.load(file)
        except (json.JSONDecodeError, IOError):
            print("Error reading tasks file. Starting with empty list.")
            return []
    return []


def save_tasks(tasks):
    """Save the tasks list to the JSON file."""
    try:
        with open(TASKS_FILE, "w") as file:
            json.dump(tasks, file, indent=2)
        print("Tasks saved successfully.")
    except IOError as e:
        print(f"Error saving tasks: {e}")


def add_task(tasks):
    """Add a new task to the list."""
    title = input("Task title: ").strip()

    # Error handling: check for empty title
    if not title:
        print("Error: Task title cannot be empty.")
        return

    new_task = {"title": title, "done": False}
    tasks.append(new_task)
    print("Task added.")
    save_tasks(tasks)


def view_tasks(tasks):
    """Display all tasks with their status."""
    if not tasks:
        print("No tasks found.")
        return

    for i, task in enumerate(tasks, 1):
        status = "Done" if task["done"] else "Pending"
        print(f"{i}. {task['title']} [{status}]")


def complete_task(tasks):
    """Mark a task as completed."""
    if not tasks:
        print("No tasks to complete.")
        return

    view_tasks(tasks)

    try:
        task_number = int(input("Task number: "))

        # Error handling: check for valid task number
        if task_number < 1 or task_number > len(tasks):
            print("Error: Invalid task number.")
            return

        tasks[task_number - 1]["done"] = True
        print("Task completed.")
        save_tasks(tasks)
    except ValueError:
        print("Error: Please enter a valid number.")


def delete_task(tasks):
    """Delete a task from the list."""
    if not tasks:
        print("No tasks to delete.")
        return

    view_tasks(tasks)

    try:
        task_number = int(input("Task number: "))

        # Error handling: check for valid task number
        if task_number < 1 or task_number > len(tasks):
            print("Error: Invalid task number.")
            return

        deleted_task = tasks.pop(task_number - 1)
        print(f"Task '{deleted_task['title']}' deleted.")
        save_tasks(tasks)
    except ValueError:
        print("Error: Please enter a valid number.")


def menu(tasks):
    """Display the menu and handle user choices."""
    while True:
        print("\n--- TASK TRACKER ---")
        print("1. Add 2. View 3. Complete 4. Delete 5. Exit")
        choice = input("Choice: ").strip()

        if choice == "1":
            add_task(tasks)
        elif choice == "2":
            view_tasks(tasks)
        elif choice == "3":
            complete_task(tasks)
        elif choice == "4":
            delete_task(tasks)
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Error: Invalid choice. Please select 1-5.")


def main():
    """Main function to start the Task Tracker."""
    print("Loading tasks...")
    tasks = load_tasks()
    if tasks:
        print(f"Loaded {len(tasks)} task(s).")
    else:
        print("No existing tasks. Starting fresh.")

    menu(tasks)


if __name__ == "__main__":
    main()
```
